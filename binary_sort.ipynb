```python
def firstSearch(arr, k):
    low = 0
    high = len(arr) - 1
    ans = -1

    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] == k:
            ans = mid         
            high = mid - 1   
        elif arr[mid] < k:
            low = mid + 1      
        else:
            high = mid - 1     

    return ans

# Driver Code
arr = [1,2,3,4,5]
k = 4

print("first", firstSearch(arr, k))
    
       
```
